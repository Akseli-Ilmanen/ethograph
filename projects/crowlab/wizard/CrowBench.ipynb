## CrowBench: parameters

```python
# Where this session's .ethograph/ (alignment, labels, settings) goes. Media folders are
# listed per source below; on a new session change session_dir and every folder.
session_dir = r"C:\Users\aksel\Documents\VidData\20241218_01_Crow1"

rig = {
    "rig_name": 'CrowBench',
    "mode": 'pair',
    "timing": 'none',
    "sources": [
        {"stream": 'video', "folder": r"C:\Users\aksel\Documents\VidData\20241218_01_Crow1", "pattern": r"2024\-12\-18_(?P<trial>[^_]+)_Crow1\-(?P<camera>.+)", "n_devices": 2},
        # {"stream": 'pose', "folder": r"C:\Users\aksel\Documents\CrowBench\20241218_01_Crow1\dlc", "pattern": r"2024\-12\-17_(?P<trial>[^_]+)_Crow1\-(?P<camera>[^D]+)DLC_resnet50_Felix_cross_SessionsAug1shuffle1_200000_filtered", "n_devices": 2, "software": 'DeepLabCut', "extension": '.csv'},
    ],
}
```

## Discover files

```python
from pathlib import Path

import pandas as pd

import ethograph as eto

# Session-wide files are not per trial; they are paired in the write cell.
sources = [
    eto.SourceSpec(
        stream=s["stream"],
        folder=s["folder"],
        pattern=s.get("pattern"),
        software=s.get("software"),
        extension=s.get("extension"),
    )
    for s in rig["sources"]
    if not s.get("session_wide")
]
table = eto.discover_media(sources)  # one row per trial, each cell a file's full path
# Optional: join per-trial metadata (a CSV/TSV with a "trial" column) onto the table.
# Delete this block if you have none; the media pairing above is all the write cell needs.
if rig.get("trial_table_file"):
    _trial_table_path = Path(rig["trial_table_file"])
    _sep = "	" if _trial_table_path.suffix == ".tsv" else ","
    trials = pd.read_csv(_trial_table_path, sep=_sep)
    table = table.merge(trials, on="trial", how="left")
table
```

## Write

```python
eto.pair_media(
    table,
    stream_rates={},
    output_path=f"{session_dir}/.ethograph/alignment.nwb",
)
```
